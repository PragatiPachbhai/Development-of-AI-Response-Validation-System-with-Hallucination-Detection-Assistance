## 1. Python Fundamentals

### Q1. What is Python?
**Answer:** Python is a high-level, general-purpose programming language known for its readable syntax. It supports procedural, object-oriented, and functional programming.

### Q2. What are some features of Python?
**Answer:** Python has readable syntax, dynamic typing, automatic memory management, a large standard library, and a broad ecosystem of third-party packages.

### Q3. What is a variable?
**Answer:** A variable is a name that refers to an object. Python determines the object's type at runtime.

```python
name = "Pragati"
age = 21
cgpa = 9.1
is_student = True

print(type(name))
print(type(age))
print(type(cgpa))
print(type(is_student))
```

### Q4. What are common built-in data types?
**Answer:** Common types include `int`, `float`, `complex`, `bool`, `str`, `list`, `tuple`, `set`, `dict`, and `NoneType`.

```python
integer_value = 10
float_value = 3.14
text_value = "Python"
list_value = [1, 2, 3]
tuple_value = (1, 2, 3)
set_value = {1, 2, 3}
dict_value = {"name": "Aman"}
nothing = None

for value in [integer_value, float_value, text_value, list_value,
              tuple_value, set_value, dict_value, nothing]:
    print(type(value))
```

### Q5. What is the difference between mutable and immutable objects?
**Answer:** Mutable objects can be changed after creation; immutable objects cannot be changed in place. Lists, dictionaries, and sets are mutable. Integers, floats, strings, and tuples are immutable as containers. A tuple can still contain a mutable object, such as a list.

```python
numbers = [10, 20, 30]
numbers[0] = 100
print(numbers)

text = "hello"
# text[0] = "H"  # Raises TypeError because strings are immutable
text = "H" + text[1:]
print(text)

mixed = ([1, 2], "hello")
mixed[0].append(3)  # The tuple is immutable, but its list item is mutable
print(mixed)
```

### Q6. What is the difference between `==` and `is`?
**Answer:** `==` compares values for equality. `is` checks whether two references point to the same object. Use `is None` when checking for `None`.

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)

result = None
if result is None:
    print("No result")
```

### Q7. What is type conversion?
**Answer:** Type conversion changes a value from one type to another, for example using `int()`, `float()`, or `str()`. Some conversions can fail if the input is invalid.

```python
x = "100"
y = int(x)
print(y + 20)
print(float("3.5"))
print(str(25))
```

## 2. Strings

### Q8. What is a string?
**Answer:** A string is an immutable sequence of Unicode characters.

```python
text = "python programming"

print(len(text))
print(text.upper())
print(text.lower())
print(text.capitalize())
print(text.count("m"))
print(text.find("pro"))
```

### Q9. What is string slicing?
**Answer:** Slicing extracts part of a sequence using `sequence[start:stop:step]`. The `stop` index is excluded.

```python
text = "Python"

print(text[0:3])
print(text[2:])
print(text[::-1])
```

### Q10. How do you check whether a string is a palindrome?
**Answer:** A palindrome reads the same forward and backward. This example ignores letter case but does not remove spaces or punctuation.

```python
text = "Level"
print(text.lower() == text.lower()[::-1])
```

## 3. Lists, Tuples, and Sets

### Q11. What is a list?
**Answer:** A list is an ordered, mutable collection that allows duplicate values.

```python
numbers = [10, 20, 30, 20]
numbers.append(40)
numbers.remove(20)  # Removes the first matching value
print(numbers)
```

### Q12. How do you find the largest element in a list?
**Answer:** Iterate through the list while keeping track of the largest value seen so far. This solution takes O(n) time.

```python
numbers = [12, 45, 7, 89, 23]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)
```

### Q13. What is a tuple?
**Answer:** A tuple is an ordered, immutable sequence. It can contain duplicate values and values of different types.

```python
point = (10, 20)
print(point[0])
print(point[1])
```

### Q14. What is a set?
**Answer:** A set is an unordered collection of unique hashable elements. It is useful for removing duplicates and performing set operations.

```python
numbers = [1, 2, 2, 3, 3, 4]
print(sorted(set(numbers)))

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a & b)  # Intersection
print(sorted(a | b))  # Union, sorted for predictable display
print(a - b)  # Difference
```

## 4. Dictionaries

### Q15. What is a dictionary?
**Answer:** A dictionary stores key-value pairs. Keys must be hashable and unique; values can be of different types. Dictionaries preserve insertion order in modern Python.

```python
student = {
    "name": "Aman",
    "age": 21,
    "marks": 85
}

print(student["name"])
student["marks"] = 90
student["city"] = "Pune"
print(student)
```

### Q16. How do you count the frequency of elements?
**Answer:** Use a dictionary to store each element as a key and its count as the value.

```python
numbers = [1, 2, 2, 3, 1, 2]
frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

## 5. Functions

### Q17. What is a function?
**Answer:** A function is a reusable block of code defined with `def`. It can accept parameters and return a result using `return`.

```python
def add(a, b):
    return a + b

print(add(5, 3))
```

### Q18. What is the difference between `print()` and `return`?
**Answer:** `print()` displays information to the console. `return` sends a value back to the caller and ends the function execution.

```python
def show_sum(a, b):
    print(a + b)

def get_sum(a, b):
    return a + b

show_sum(2, 3)
result = get_sum(2, 3)
print(result * 2)
```

### Q19. What is a lambda function?
**Answer:** A lambda is a small anonymous function consisting of one expression.

```python
square = lambda x: x * x
print(square(5))
```

## 6. Object-Oriented Programming

### Q20. What are a class and an object?
**Answer:** A class defines the attributes and behavior of a type. An object is an instance of a class.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(self.name, self.marks)

student = Student("Aman", 85)
student.display()
```

### Q21. What is inheritance?
**Answer:** Inheritance allows a class (child class) to reuse or extend the attributes and methods of another class (parent class).

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

dog = Dog()
dog.speak()
```

### Q22. What is polymorphism?
**Answer:** Polymorphism means that different object types can provide their own implementations of a common method or interface.

```python
class Cat:
    def speak(self):
        print("Meow")

class Dog:
    def speak(self):
        print("Bark")

for animal in [Cat(), Dog()]:
    animal.speak()
```

## 7. Exception Handling

### Q23. What is exception handling?
**Answer:** Exception handling manages runtime errors using `try` and `except`. `else` runs if no exception occurs, and `finally` runs whether or not an exception occurs.

```python
try:
    number = int("123")
    result = 10 / number
except ValueError:
    print("Invalid number")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print(result)
finally:
    print("Finished")
```

## 8. NumPy

### Q24. What is NumPy?
**Answer:** NumPy is a Python library for numerical computing. Its main structure is the multidimensional `ndarray`, which supports efficient array operations.

```python
import numpy as np

python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print(python_list * 2)
print(numpy_array * 2)
```

### Q25. What do `shape`, `ndim`, `size`, and `dtype` mean?
**Answer:**
- `shape`: length of the array along each dimension.
- `ndim`: number of dimensions.
- `size`: total number of elements.
- `dtype`: data type of the array elements.

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

### Q26. What is vectorization?
**Answer:** Vectorization applies an operation to an array's elements without writing an explicit Python loop.

```python
import numpy as np

arr = np.array([10, 20, 30, 40])
print(arr + 5)
```

### Q27. Name some useful NumPy functions.
**Answer:** `np.array()`, `np.zeros()`, `np.ones()`, `np.arange()`, `np.linspace()`, `np.reshape()` (or an array's `.reshape()` method), `np.concatenate()`, `np.mean()`, and `np.sum()`.

```python
import numpy as np

arr = np.arange(1, 7)
print(arr.reshape(2, 3))
print(np.mean(arr))
print(np.sum(arr))
```

## 9. Pandas

### Q28. What is Pandas?
**Answer:** Pandas is a Python library for working with labeled and tabular data. Its main data structures are `Series` (one-dimensional) and `DataFrame` (two-dimensional).

```python
import pandas as pd

data = {
    "Name": ["Aman", "Sara", "Rahul"],
    "Age": [21, 22, 20],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)
print(df)
```

### Q29. What are `head()`, `tail()`, `info()`, and `describe()`?
**Answer:**
- `head()`: returns the first five rows by default.
- `tail()`: returns the last five rows by default.
- `info()`: prints column names, data types, and non-null counts.
- `describe()`: summarizes numeric columns by default.

```python
print(df.head(2))
print(df.tail(2))
df.info()
print(df.describe())
```

### Q30. How do you filter rows in Pandas?
**Answer:** Use a Boolean condition to select rows that meet the condition.

```python
print(df[df["Marks"] >= 85])
```

### Q31. How do you handle missing values in Pandas?
**Answer:** Use `isna()` or `isnull()` to identify missing values, `dropna()` to remove rows or columns, and `fillna()` to replace missing values. Choose the approach based on the data and task.

```python
import pandas as pd

data = {"Name": ["Aman", "Sara", "Rahul"], "Marks": [85, None, 78]}
df_missing = pd.DataFrame(data)

print(df_missing.isna())
df_missing["Marks"] = df_missing["Marks"].fillna(df_missing["Marks"].mean())
print(df_missing)
```

## 11. Common Coding Questions

### Q35. Check whether a number is even or odd.

```python
number = 7

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

### Q36. Find the factorial of a number.
**Answer:** The factorial of a non-negative integer `n` is the product of all integers from 1 to `n`. By convention, `0! = 1`.

```python
n = 5
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(factorial)
```

### Q37. Check whether a number is prime.
**Answer:** A prime number is an integer greater than 1 with exactly two positive divisors: 1 and itself. Checking divisors only up to the square root is sufficient.

```python
n = 29
is_prime = n > 1

for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        is_prime = False
        break

print(is_prime)
```

### Q38. Generate the Fibonacci sequence.
**Answer:** Each number after the first two is the sum of the previous two numbers.

```python
n = 7
a, b = 0, 1
sequence = []

for _ in range(n):
    sequence.append(a)
    a, b = b, a + b

print(sequence)
```

### Q39. Find the second-largest distinct element in a list.
**Answer:** Convert to a set to remove duplicates, sort the values, and select the second-last value. This approach takes O(n log n) time because of sorting. It requires at least two distinct values.

```python
numbers = [10, 5, 8, 10, 8, 3]
distinct_numbers = sorted(set(numbers))

if len(distinct_numbers) >= 2:
    print(distinct_numbers[-2])
else:
    print("No second-largest distinct value")
```

### Q40. Count vowels in a string.

```python
text = "Python Interview"
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1

print(count)
```
